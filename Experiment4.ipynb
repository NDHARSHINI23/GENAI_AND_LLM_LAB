```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

# Load pretrained model
model_name = "google/flan-t5-large"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Chatbot: Hello! Type 'exit' to stop.\n")

while True:

    user = input("You: ")

    # Exit
    if user.lower() == "exit":
        print("Chatbot: Goodbye! Have a nice day.")
        break

    # Common greetings
    elif user.lower() in ["hi", "hello", "hey"]:
        print("Chatbot: Hello! How can I help you today?\n")
        continue

    # AI
    elif "artificial intelligence" in user.lower():
        print("Chatbot: Artificial Intelligence (AI) is the simulation of human intelligence by machines to perform tasks such as learning, reasoning, problem-solving, and decision-making.\n")
        continue

    # LLM
    elif "large language model" in user.lower() or "large language models" in user.lower():
        print("Chatbot: Large Language Models (LLMs) are transformer-based AI models trained on massive text datasets to understand and generate human language. They are used for text generation, summarization, translation, and question answering.\n")
        continue

    # ML
    elif "machine learning" in user.lower():
        print("Chatbot: Machine Learning is a branch of Artificial Intelligence that enables computers to learn from data and improve their performance without being explicitly programmed.\n")
        continue

    # DL
    elif "deep learning" in user.lower():
        print("Chatbot: Deep Learning is a subset of Machine Learning that uses multi-layer neural networks to learn complex patterns from data.\n")
        continue

    # Thank you
    elif "thank" in user.lower():
        print("Chatbot: You're welcome! Happy learning.\n")
        continue

    # Otherwise use pretrained model
    prompt = f"""
You are a helpful AI assistant.

Answer the following question clearly in 2-3 sentences.

Question:
{user}

Answer:
"""

    inputs = tokenizer(prompt, return_tensors="pt")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=120,
            temperature=0.7,
            do_sample=True
        )

    answer = tokenizer.decode(outputs[0], skip_special_tokens=True)

    print("Chatbot:", answer)
    print()
```
