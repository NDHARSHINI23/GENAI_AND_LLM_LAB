```python
# 1. Install dependencies
!pip install transformers torch accelerate sentencepiece -q

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# 2. Load Instruction-Tuned Code Model
model_name = "Qwen/Qwen2.5-Coder-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=(
        torch.float16
        if torch.cuda.is_available()
        else torch.float32
    ),
    device_map="auto"
)


# 3. Response Generation Function
def generate_response(prompt, max_tokens=350):

    messages = [
        {
            "role": "system",
            "content": (
                "You are an expert Python programming assistant. "
                "Follow the user's instructions exactly. "
                "Give clear, correct and properly formatted answers."
            )
        },
        {
            "role": "user",
            "content": prompt
        }
    ]

    formatted_prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        formatted_prompt,
        return_tensors="pt"
    ).to(model.device)

    outputs = model.generate(
        **inputs,
        max_new_tokens=max_tokens,
        do_sample=False
    )

    generated_tokens = outputs[0][
        inputs.input_ids.shape[1]:
    ]

    response = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return response.strip()


# 4. Code Generation Task
code_task = """
Write a Python program to calculate the factorial of a number
using a recursive function.

Requirements:
1. Accept a number from the user.
2. Use recursion.
3. Display the factorial.
4. Handle negative input.

Return only the complete Python program.
Do not include explanations or Markdown code fences.
"""

generated_code = generate_response(
    f"Generate Python code for the following task:\n{code_task}",
    max_tokens=250
)


# 5. Debugging Assistant Task
# Intentionally contains an error: len(number)
f_code = """
def calculate_average(numbers):
    total = sum(numbers)
    average = total / len(number)
    return average

values = [10, 20, 30, 40]
print("Average:", calculate_average(values))
"""


debugging_prompt = f"""
Analyze the following Python program.

Code:
{f_code}

There is a programming error in this code.

Tasks:
1. Identify the exact error.
2. Explain the cause briefly.
3. Provide the corrected Python program.
4. Provide the corrected program output.

Use EXACTLY this format:

Error:
<identify the error>

Explanation:
<brief explanation>

Corrected Code:
<complete corrected Python program>

Corrected Program Output:
<output>

Important:
- The function parameter is "numbers".
- The incorrect code uses len(number).
- The correct expression is len(numbers).
- Do not invent another error.
"""

debugging_result = generate_response(
    debugging_prompt,
    max_tokens=300
)


# 6. Display Results
print("AI-POWERED CODE GENERATION AND DEBUGGING ASSISTANT")
print("=" * 60)

print("\n1. GENERATED CODE")
print("-" * 60)
print(generated_code)

print("\n2. DEBUGGING RESULT")
print("-" * 60)
print(debugging_result)
```
